# обрабтка выбросов

```python
import pandas as pd
import numpy as np

df = pd.read_csv('dataset/cleaned_dataset.csv')
print(df.shape)
df.describe()
```

```python
n_before = len(df)
print(f"До обрезки: {n_before} строк")

# === Цена: квантили 1% и 99% ===
price_low = df['Price'].quantile(0.01)
price_high = df['Price'].quantile(0.99)
print(f"\nЦена:")
print(f"  Квантиль 1%:  {price_low:>15,.0f} ₽")
print(f"  Квантиль 99%: {price_high:>15,.0f} ₽")
print(f"  Удалим объекты дешевле {price_low:,.0f} ₽ и дороже {price_high:,.0f} ₽")

# === Площадь: квантили 1% и 99% ===
area_low = df['Area'].quantile(0.01)
area_high = df['Area'].quantile(0.99)
print(f"\nПлощадь:")
print(f"  Квантиль 1%:  {area_low:>6.1f} м²")
print(f"  Квантиль 99%: {area_high:>6.1f} м²")
print(f"  Удалим объекты меньше {area_low:.1f} м² и больше {area_high:.1f} м²")

# Применяем фильтры
mask = (
    (df['Price']  >= price_low)  & (df['Price']  <= price_high) &
    (df['Area']   >= area_low)   & (df['Area']   <= area_high)
)
df = df[mask].reset_index(drop=True)

n_after = len(df)
print(f"\nПосле обрезки: {n_after} строк")
print(f"Удалено: {n_before - n_after} строк ({(n_before-n_after)/n_before*100:.2f}%)")

# Контрольная статистика после обрезки
print("\n=== Цена после обрезки ===")
print(df['Price'].describe().apply(lambda x: f'{x:,.0f}'))
print("\n=== Площадь после обрезки ===")
print(df['Area'].describe().apply(lambda x: f'{x:,.1f}'))
```

```python
df.to_csv('dataset/data_third_version.csv', index=False)
```
